```python
import matplotlib.pyplot as plt

factors = [
    'Погода и давление', 'Сезон', 'Нерест',
    'Время суток', 'Фаза Луны',
    'Место ловли', 'Снасти'
]
weights = [30, 20, 15, 15, 5, 10, 5]
plt.figure(figsize=(8, 8))
plt.pie(weights, labels=factors, autopct='%1.0f%%', colors=plt.cm.Paired.colors)
plt.title("Веса факторов, влияющих на активность клёва", fontsize=14)
plt.show()
```
