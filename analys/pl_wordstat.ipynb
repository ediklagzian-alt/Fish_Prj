```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/wordstat_dynamic.csv")

plt.figure(figsize=(12, 6))
plt.plot(df["Период"], df["Число запросов"], marker='o', color='blue')

plt.title("Сезонность спроса на прогнозы клёва (Яндекс Wordstat)", fontsize=14)
plt.xlabel("Месяц", fontsize=12)
plt.ylabel("Количество запросов", fontsize=12)
plt.xticks(rotation=45)
plt.grid(True, linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
